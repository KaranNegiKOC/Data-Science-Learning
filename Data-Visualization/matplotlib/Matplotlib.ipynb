```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

**Matplotlib Basics**

```python
x = np.linspace(0,5,11)
x 
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.ylabel("y label")
plt.xlabel("X label")
```

**Subplots in Matplotlib**

```python
plt.subplot(2,2,1)
plt.subplot(2,2,4)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,2)
plt.plot(x**2, x**2)
```

**object oriented way**

```python
fig= plt.figure(figsize = (6,8), dpi = 100)
axis1= fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.5,0.3,0.3])
axis2.plot(x,y)
```

**Types of plot**

```python
plt.scatter(x,y)
```

```python
from random import sample
data = sample(range(1, 1000), 100)
plt.hist(data)
```

```python
plt.boxplot(x)
```

**Saving plots**

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basic plot.png")
```
